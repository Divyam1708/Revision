```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]
base=int(input("Base: "))
power=int(input('Power: '))
c=0
c1=0
for x in ((adj+fruits)*base**power):
    if(c%(len(adj+fruits))==0):
        c1+=1
        print(c1)
    c=c+1
    print(x)
```
